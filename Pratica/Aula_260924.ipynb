```python
CAMINHO_VENDAS= r"C:\_Pessoas\Bruno.Araujo\CDN_IFAD2\Pratica\vendas_spark.csv"
CAMINHO_CLIENTES = r"C:\_Pessoas\Bruno.Araujo\CDN_IFAD2\Pratica\clientes_spark.csv"
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Exemplo_Spark_Drive")  \
    .master("local[*]") \
    .getOrCreate()
    
print(f"Spark iniciado com sucesso: {spark.version}")
```

```python
df_vendas = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(CAMINHO_VENDAS)
```

```python
df_vendas.printSchema()
```

```python
df_vendas.show(5)
```

```python
print(f"O df_vendas possui {df_vendas.count()}")
```

```python
df_clientes = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(CAMINHO_CLIENTES)    
```

```python
df_clientes.printSchema()
```

```python
df_clientes.show(5)
```

```python
print(f"O df_clientes possui {df_clientes.count()}")
```
