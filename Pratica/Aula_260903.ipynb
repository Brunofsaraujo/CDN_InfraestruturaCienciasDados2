```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Exemplo_Spark") \
    .master("local[*]") \
    .getOrCreate()
    
print(f"Spark iniciado com sucesso V-{spark.version}")
```

```python
colunas = [
    "id_venda",
    "produto",
    "categoria",
    "estado",
    "quantidade",
    "preco_unitario"
]

dados_vendas = [
    (1, "Notebook", "Informatica", "SP", 13, 6000.00),
    (2, "Mouse", "Informatica", "SP", 13, 30.00),
    (3, "Teclado", "Informatica", "SP", 13, 50.00),
    (4, "Monitor", "Informatica", "SP", 50, 1200.00),
    (5, "Cadeira", "Móveis", "SP", 50, 80.00),
    (6, "Mesa", "Móveis", "SP", 50, 200.00),
    (7, "Armário", "Móveis", "SP", 3, 3000.00),
    (8, "Televisor", "Eletrônicos", "SP", 5, 2500.00),
    (9, "Ar Condicionado", "Eletrônicos", "SP", 5, 5000.00),
    (10, "Ventilador", "Eletricos", "SP", 5, 220.00)
]

df_vendas = spark.createDataFrame(dados_vendas, colunas)
```

```python
from pyspark.sql.functions import col

df_vendas = df_vendas.withColumn(
    "valor_total",
    col("quantidade") * col("preco_unitario")
)
```

```python
df_vendas.show()
```

```python
# GROUP BY

from pyspark.sql.functions import sum

vendas_por_categoria = df_vendas.groupBy(
    "categoria"
).agg(
    sum("valor_total")
    .alias("total_categorizado")
)
```

```python
vendas_por_categoria.show()
```

```python
from pyspark.sql.functions import(
    sum,
    count,
    avg,
    round
)

df_vendas_valores = df_vendas.groupBy(
    "categoria"
).agg(
    count("id_venda").alias("quantidade"),
    sum("quantidade").alias("qtde_vendida"),
    sum("valor_total").alias("total_vendas"),
    round(avg("valor_total"), 2).alias("media_venda")
)   
```

```python
df_vendas_valores.show()
```

```python
# JOIN = Juncao de tabelas (data frame)

dados_clientes = [
    (1, "Mario", "SP"),
    (2, "Luigi", "RJ"),
    (3, "Chaves", "MG"),
    (4, "Kiko", "RS"),
    (5, "Sr. Madruga", "SP"),
]

colunas_clientes = ["id_cliente", "nome", "uf"]

df_clientes = spark.createDataFrame(
    dados_clientes,
    colunas_clientes
)

df_clientes.show()
```

```python
# (1, "Notebook", "Informatica", "SP", 13, 6000.00),
# (2, "Mouse", "Informatica", "SP", 13, 30.00),
# (3, "Teclado", "Informatica", "SP", 13, 50.00),
# (4, "Monitor", "Informatica", "SP", 50, 1200.00),
# (5, "Cadeira", "Móveis", "SP", 50, 80.00),
# (6, "Mesa", "Móveis", "SP", 50, 200.00),
# (7, "Armário", "Móveis", "SP", 3, 3000.00),
# (8, "Televisor", "Eletrônicos", "SP", 5, 2500.00),
# (9, "Ar Condicionado", "Eletrônicos", "SP", 5, 5000.00),
# (10, "Ventilador", "Eletricos", "SP", 5, 220.00)

dados_vendas_clientes_colunas = [
    "id_venda_clientes",
    "id_venda",
    "id_cliente"
]

dados_vendas_clientes = [
    (1, 1, 1),
    (2, 2, 2),
    (3, 3, 3),
    (4, 4, 5),
    (5, 6, 5),
    (6, 7, 5),
    (7, 8, 2),  
    (8, 9, 3),
    (9, 10, 5)
]

df_vendas_clientes = spark.createDataFrame(
    dados_vendas_clientes,
    dados_vendas_clientes_colunas    
)

df_vendas_clientes.show()
```

```python
# primeiro join

resultado_parcial = df_vendas_clientes.join(
    df_clientes,
    df_vendas_clientes.id_cliente == df_clientes.id_cliente,
    "inner"
)

# segundo join

resultado_final = resultado_parcial.join(
    df_vendas,
    resultado_parcial.id_venda == df_vendas.id_venda,
    "inner"
)

resultado_final.show()
```
